### Missing Value Check

There are 89 neighborhood records with missing values for `total_bedrooms`. All 89 of these records have a postal code available, so the missing bedroom values can be imputed using the average `total_bedrooms` value for the corresponding postal code as required by the assignment.

There are also 11 records with a missing postal code, but none of the records with missing bedrooms are missing their postal code. This means the postal code based bedroom imputation can be completed without dropping any of the records that have missing bedroom values.

The next step is to calculate the average number of bedrooms for each postal code, fill the missing bedroom values, and then calculate bedrooms per household.

```python
# Impute missing bedroom values using the postal code average
try:
    # Calculate the average total bedrooms for each postal code
    postal_code_bedroom_avg = neighborhood_df.groupby(
        "postal_code"
    )["total_bedrooms"].transform("mean")
    # Fill missing bedroom values with the average for the postal code
    neighborhood_df["total_bedrooms_imputed"] = (
        neighborhood_df["total_bedrooms"].fillna(
            postal_code_bedroom_avg
        )
    )
    # Calculate bedrooms per household using the imputed values
    neighborhood_df["bedrooms_per_household"] = (
        neighborhood_df["total_bedrooms_imputed"]
        / neighborhood_df["households"]
    )
    print("Bedroom imputation completed successfully.")
    print(
        "\nMissing total bedrooms before imputation:",
        neighborhood_df["total_bedrooms"].isnull().sum()
    )
    print(
        "Missing total bedrooms after imputation:",
        neighborhood_df[
            "total_bedrooms_imputed"
        ].isnull().sum()
    )
    print(
        "Missing bedrooms per household:",
        neighborhood_df[
            "bedrooms_per_household"
        ].isnull().sum()
    )
    # Show some of the records that originally had missing bedrooms
    print("\nSample imputed records:")
    display(
        neighborhood_df.loc[
            neighborhood_df["total_bedrooms"].isnull(),
            [
                "neighborhood-political",
                "postal_code",
                "total_bedrooms",
                "total_bedrooms_imputed",
                "households",
                "bedrooms_per_household"
            ]
        ].head(10)
    )
except Exception as e:
    print(f"Bedroom imputation failed: {e}")
    raise
```

### Postal Code Imputation Check

The complete merged dataset contains 26 housing records for postal code 93940. Of these records, 25 have a valid `total_bedrooms` value, producing an average of 514.32 bedrooms for the postal code.

This confirms that the remaining missing value for Jack's Peak can be imputed using the postal code average as required. The earlier missing value occurred because the average was calculated only from records that already had neighborhood information. To avoid this problem, I will calculate postal code bedroom averages from the complete merged dataset and use those averages to fill missing bedroom values in the neighborhood records.

```python
# Impute missing bedrooms using postal code averages from the full dataset
try:
    # Calculate average total bedrooms for each postal code
    # using all available records in the merged dataset
    postal_code_bedroom_avg = (
        housing_gmaps_df.groupby("postal_code")["total_bedrooms"]
        .mean()
    )
    # Keep the original bedroom values in a separate column
    neighborhood_df["total_bedrooms_imputed"] = (
        neighborhood_df["total_bedrooms"]
    )
    # Find the records that require imputation
    missing_bedrooms_mask = (
        neighborhood_df["total_bedrooms_imputed"].isnull()
    )
    # Fill missing bedrooms using the corresponding postal code average
    neighborhood_df.loc[
        missing_bedrooms_mask,
        "total_bedrooms_imputed"
    ] = (
        neighborhood_df.loc[
            missing_bedrooms_mask,
            "postal_code"
        ].map(postal_code_bedroom_avg)
    )
    # Calculate bedrooms per household
    neighborhood_df["bedrooms_per_household"] = (
        neighborhood_df["total_bedrooms_imputed"]
        / neighborhood_df["households"]
    )
    print("Final bedroom imputation completed successfully.")
    print(
        "\nRecords originally missing total bedrooms:",
        neighborhood_df["total_bedrooms"].isnull().sum()
    )
    print(
        "Missing bedrooms after imputation:",
        neighborhood_df[
            "total_bedrooms_imputed"
        ].isnull().sum()
    )
    print(
        "Missing bedrooms per household:",
        neighborhood_df[
            "bedrooms_per_household"
        ].isnull().sum()
    )
    # Verify the records that were imputed
    print("\nSample imputed records:")
    display(
        neighborhood_df.loc[
            neighborhood_df["total_bedrooms"].isnull(),
            [
                "neighborhood-political",
                "postal_code",
                "total_bedrooms",
                "total_bedrooms_imputed",
                "households",
                "bedrooms_per_household"
            ]
        ].head(10)
    )
    # Verify the Jack's Peak record specifically
    print("\nJack's Peak verification:")
    display(
        neighborhood_df[
            neighborhood_df["neighborhood-political"]
            == "Jack's Peak"
        ][
            [
                "neighborhood-political",
                "postal_code",
                "total_bedrooms",
                "total_bedrooms_imputed",
                "households",
                "bedrooms_per_household"
            ]
        ]
    )
except Exception as e:
    print(f"Final bedroom imputation failed: {e}")
    raise
```

The data is now ready for the remaining neighborhood level feature engineering and aggregation.

```python
# Create ocean proximity indicators
# Create one hot encoded ocean proximity features
try:
    # Check the ocean proximity categories before encoding
    print("Ocean proximity categories:")
    print(
        neighborhood_df["ocean_proximity"]
        .value_counts()
    )
    # Create the five required ocean proximity features
    neighborhood_df["less_than_1h_ocean"] = (
        neighborhood_df["ocean_proximity"] == "<1H OCEAN"
    ).astype(int)
    neighborhood_df["inland"] = (
        neighborhood_df["ocean_proximity"] == "INLAND"
    ).astype(int)
    neighborhood_df["island"] = (
        neighborhood_df["ocean_proximity"] == "ISLAND"
    ).astype(int)
    neighborhood_df["near_bay"] = (
        neighborhood_df["ocean_proximity"] == "NEAR BAY"
    ).astype(int)
    neighborhood_df["near_ocean"] = (
        neighborhood_df["ocean_proximity"] == "NEAR OCEAN"
    ).astype(int)
    print("\nOcean proximity features created successfully.")
    # Verify the one hot encoded features
    print("\nFeature totals:")
    print(
        neighborhood_df[
            [
                "less_than_1h_ocean",
                "inland",
                "island",
                "near_bay",
                "near_ocean"
            ]
        ].sum()
    )
    print("\nSample encoded records:")
    display(
        neighborhood_df[
            [
                "neighborhood-political",
                "ocean_proximity",
                "less_than_1h_ocean",
                "inland",
                "island",
                "near_bay",
                "near_ocean"
            ]
        ].head(10)
    )
except Exception as e:
    print(f"Ocean proximity encoding failed: {e}")
    raise
```

### Ocean Proximity Features

The `ocean_proximity` field was converted into the five one hot encoded features required for the neighborhood feature group. Each record now indicates whether it is less than one hour from the ocean, inland, on an island, near the bay, or near the ocean.

The encoded feature totals account for all 9,000 neighborhood records. There are no island records in the neighborhood subset, so the `island` feature contains zeros for all records. The feature is still retained because it is part of the required feature group schema.

The next step is to aggregate the housing records by neighborhood and calculate the required neighborhood level values.

```python
# Create one record for each neighborhood
import numpy as np
try:
    # Aggregate the required values by neighborhood
    neighborhood_features_df = (
        neighborhood_df.groupby(
            "neighborhood-political",
            as_index=False
        )
        .agg(
            less_than_1h_ocean=("less_than_1h_ocean", "max"),
            inland=("inland", "max"),
            island=("island", "max"),
            near_bay=("near_bay", "max"),
            near_ocean=("near_ocean", "max"),
            median_house_value=("median_house_value", "mean"),
            median_house_age=("housing_median_age", "mean"),
            total_households=("households", "mean"),
            bedrooms_per_household=("bedrooms_per_household", "mean")
        )
    )
    # Rename the neighborhood field for the Feature Store primary key
    neighborhood_features_df = neighborhood_features_df.rename(
        columns={
            "neighborhood-political": "neighborhood"
        }
    )
    # Cap the average median house value at 500,000
    neighborhood_features_df["median_house_value"] = (
        neighborhood_features_df["median_house_value"]
        .clip(upper=500000)
    )
    # Discretize average house age into groups of 10 years
    neighborhood_features_df["median_house_age"] = (
        np.floor(
            neighborhood_features_df["median_house_age"] / 10
        ) * 10
    ).astype(int)
    # Round the average number of households up to an integer
    neighborhood_features_df["total_households"] = (
        np.ceil(
            neighborhood_features_df["total_households"]
        )
    ).astype(int)
    print("Neighborhood features created successfully.")
    print(
        "\nNeighborhood feature group shape:",
        neighborhood_features_df.shape
    )
    print(
        "Unique neighborhoods:",
        neighborhood_features_df["neighborhood"].nunique()
    )
    print("\nSample neighborhood features:")
    display(
        neighborhood_features_df.head(10)
    )
    # Check the three neighborhoods required by the assignment
    required_neighborhoods = [
        "Brooktree",
        "Fisherman's Wharf",
        "Los Osos"
    ]
    print("\nRequired neighborhood records:")
    display(
        neighborhood_features_df[
            neighborhood_features_df["neighborhood"].isin(
                required_neighborhoods
            )
        ]
    )
except Exception as e:
    print(f"Neighborhood feature creation failed: {e}")
    raise
```

### Neighborhood Feature Engineering Results

The housing records were aggregated into 1,306 unique neighborhoods. Each neighborhood now has the five required ocean proximity indicators along with the derived housing features.

The average median house value was capped at $500,000, the average house age was grouped into 10 year intervals, and the average number of households was rounded up to an integer. Bedrooms per household was calculated using the bedroom values after postal code based imputation.

The three neighborhoods required for the final Feature Store queries, Brooktree, Fisherman's Wharf, and Los Osos, are all present in the processed data. The next step is to prepare these neighborhood features for ingestion into SageMaker Feature Store.

```python
# Add the event time required by SageMaker Feature Store
import time
import boto3
try:
    # Feature group name used for checking existing records
    feature_group_name = "neighborhood-feature-group"
    # Create the Feature Store runtime client
    featurestore_runtime_check = boto3.client(
        "sagemaker-featurestore-runtime"
    )
    # Check whether the feature group already contains data
    try:
        existing_response = featurestore_runtime_check.get_record(
            FeatureGroupName=feature_group_name,
            RecordIdentifierValueAsString="Brooktree"
        )
        existing_record = existing_response.get("Record", [])
        if existing_record:
            # Reuse the event time from the existing Feature Store data
            event_time_feature = next(
                feature
                for feature in existing_record
                if feature["FeatureName"] == "event_time"
            )
            ingestion_time = float(
                event_time_feature["ValueAsString"]
            )
            print(
                "Existing Feature Store data found."
            )
            print(
                "Reusing event time:",
                ingestion_time
            )
        else:
            ingestion_time = float(time.time())
            print(
                "No existing Feature Store data found."
            )
            print(
                "New event time created:",
                ingestion_time
            )
    except Exception:
        # The feature group may not exist during the first run
        ingestion_time = float(time.time())
        print(
            "Feature Store data not available yet."
        )
        print(
            "New event time created:",
            ingestion_time
        )
    # Add the event time to the neighborhood features
    neighborhood_features_df["event_time"] = ingestion_time
    # Arrange the columns in the order used for the feature group
    feature_store_df = neighborhood_features_df[
        [
            "neighborhood",
            "event_time",
            "less_than_1h_ocean",
            "inland",
            "island",
            "near_bay",
            "near_ocean",
            "median_house_value",
            "median_house_age",
            "total_households",
            "bedrooms_per_household"
        ]
    ].copy()
    print("\nFeature Store data prepared successfully.")
    print(
        "\nFeature Store shape:",
        feature_store_df.shape
    )
    print("\nColumn data types:")
    print(feature_store_df.dtypes)
    print("\nMissing values:")
    print(feature_store_df.isnull().sum())
    print("\nDuplicate neighborhood keys:")
    print(
        feature_store_df["neighborhood"]
        .duplicated()
        .sum()
    )
    print("\nSample Feature Store records:")
    display(feature_store_df.head(10))
    # Verify the three neighborhoods 
    required_neighborhoods = [
        "Brooktree",
        "Fisherman's Wharf",
        "Los Osos"
    ]
    print("\nRequired neighborhood records:")
    display(
        feature_store_df[
            feature_store_df["neighborhood"].isin(
                required_neighborhoods
            )
        ]
    )
except Exception as e:
    print(f"Feature Store data preparation failed: {e}")
    raise
```

### Feature Store Data Validation

The final neighborhood dataset contains 1,306 records and 11 features, including the neighborhood primary key and the ingestion event time. The dataset contains no missing values and no duplicate neighborhood keys, so each neighborhood can be stored as a unique record in SageMaker Feature Store.

The three neighborhoods required for the assignment are also present in the final dataset. The event time was generated in Python and will be used as the Feature Store event time field during ingestion.

The processed data is now ready to be used to create the SageMaker Feature Group.

```python
# AWS and SageMaker configuration
import boto3
import sagemaker
try:
    # Create the SageMaker session
    sagemaker_session = sagemaker.Session()
    # Get the AWS region and execution role
    region = boto3.Session().region_name
    role = sagemaker.get_execution_role()
    # Create AWS clients used for Feature Store
    sagemaker_client = boto3.client(
        "sagemaker",
        region_name=region
    )
    featurestore_runtime = boto3.client(
        "sagemaker-featurestore-runtime",
        region_name=region
    )
    # Use the SageMaker default bucket for the offline store
    bucket = sagemaker_session.default_bucket()
    print("AWS configuration completed successfully.")
    print("Region:", region)
    print("S3 bucket:", bucket)
    print("Execution role:", role)
except Exception as e:
    print(f"AWS configuration failed: {e}")
    raise
```

The next step is to define the neighborhood feature group using the processed dataset.

```python
# Define the neighborhood Feature Store feature group
from sagemaker.feature_store.feature_group import FeatureGroup
try:
    # Name of the Feature Store feature group
    feature_group_name = "neighborhood-feature-group"
    # Create the FeatureGroup object
    neighborhood_feature_group = FeatureGroup(
        name=feature_group_name,
        sagemaker_session=sagemaker_session
    )
    # Load the feature definitions from the final dataframe
    neighborhood_feature_group.load_feature_definitions(
        data_frame=feature_store_df
    )
    print("Feature group definition created successfully.")
    print("Feature group name:", feature_group_name)
    print("\nFeature definitions:")
    for feature in neighborhood_feature_group.feature_definitions:
        print(
            feature.feature_name,
            "-",
            feature.feature_type
        )
except Exception as e:
    print(f"Feature group definition failed: {e}")
    raise
```

### Feature Group Definition

The neighborhood feature group schema was created successfully from the processed dataframe. SageMaker identified the neighborhood primary key as a string, the event time and integer based features as integral values, and the continuous housing features as fractional values.

The inferred feature types match the data prepared for the neighborhood feature group. The next step is to create the feature group in SageMaker Feature Store using `neighborhood` as the record identifier and `event_time` as the event time feature.

```python
# Create the neighborhood feature group if it does not already exist
from botocore.exceptions import ClientError
try:
    # S3 location used by the Feature Store offline store
    offline_store_uri = (
        f"s3://{bucket}/feature-store/neighborhood-feature-group"
    )
    # Check whether the feature group already exists
    try:
        existing_feature_group = (
            sagemaker_client.describe_feature_group(
                FeatureGroupName=feature_group_name
            )
        )
        print("Feature group already exists!!!!")
        print("Feature group:", feature_group_name)
        print(
            "Current status:",
            existing_feature_group["FeatureGroupStatus"]
        )
    except ClientError as e:
        error_code = e.response["Error"]["Code"]
        # Create the feature group only when it does not exist
        if error_code == "ResourceNotFound":
            print("Creating feature group...")
            print("Feature group:", feature_group_name)
            print("Offline store:", offline_store_uri)
            neighborhood_feature_group.create(
                s3_uri=offline_store_uri,
                record_identifier_name="neighborhood",
                event_time_feature_name="event_time",
                role_arn=role,
                enable_online_store=True
            )
            print(
                "\nFeature group creation request "
                "submitted successfully."
            )
        else:
            raise
except Exception as e:
    print(f"Feature group setup failed: {e}")
    raise
```

### Feature Group Creation

The neighborhood feature group creation request was submitted successfully to SageMaker Feature Store. The feature group uses `neighborhood` as the record identifier and `event_time` as the event time feature. The online store is enabled for retrieving individual neighborhood records, and the offline store is configured in the SageMaker S3 bucket.

Feature group creation is asynchronous, so I will verify that the feature group reaches the `Created` status before ingesting the neighborhood records.

```python
# Check the status of the neighborhood feature group
import time
try:
    while True:
        feature_group_description = (
            sagemaker_client.describe_feature_group(
                FeatureGroupName=feature_group_name
            )
        )
        feature_group_status = (
            feature_group_description["FeatureGroupStatus"]
        )
        print(
            "Feature group status:",
            feature_group_status
        )
        if feature_group_status == "Created":
            print(
                "\nFeature group is ready for data ingestion."
            )
            break
        if feature_group_status in [
            "CreateFailed",
            "DeleteFailed"
        ]:
            raise RuntimeError(
                f"Feature group entered status: "
                f"{feature_group_status}"
            )
        # Wait before checking the status again
        time.sleep(5)
except Exception as e:
    print(f"Feature group status check failed: {e}")
    raise
```
